As seen above, we implemented the brute method with linear search with a complexity of $O(N)$, but that's not the most optimized solution.

We can implement binary search.

## Steps to implementing binary search

1. Split the list `nums` at index `mid`
2. Match Check: If the middle element is equal to the `target`, return it
3. Check the right half: Is `nums[mid] < nums[hi]`? (Right side is sorted)
    * If Yes: Is `target > nums[mid]` and `target <= nums[hi]`?
        * Yes: Search right side (`lo = mid + 1`).
        * No: Search left side (`hi = mid - 1`).
4. Else (Left Half is Sorted): (nums[mid] >= nums[hi])
    * Is `target >= nums[lo]` and `target < nums[mid]`?
        * Yes: Search left side (`hi = mid - 1`).
        * No: Search right side (`lo = mid + 1`).
5. If the loop ends without finding `target`, return `-1`.
