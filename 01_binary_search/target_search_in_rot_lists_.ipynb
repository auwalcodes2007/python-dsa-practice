As seen above, we implemented the brute method with linear search with a complexity of $O(N)$, but that's not the most optimized solution.

We can implement binary search.

## Steps to implementing binary search

1. Split the list `nums` at index `mid`
2. If the middle element is equal to the `target`, return it
3. Otherwise, check if the mid number is less than the number at the highest index - `hi`
4. if yes, then everything to the right is sorted in ascending order
5. In that case, check if the target is less than or equal to the number at hi and greater than the mid number. If true, repeat. Else, the target lies on the left
6. if the mid number is greater than the number at hi, then to the left, the list is sorted in ascending order from lo to mid.
7. In that case, if target is less than mid but greater than or equal to lo, then repeat search in the left
