Let's first implement the brute force solution
1. Set `start` and `end` to `-1`
2. Loop through `nums`
3. If current number is equal to `target`, update `start`
3. If `start` is no longer `-1`, and current iteration is equal to `target`, update `end`
4. Return `start` and `end` at the end of the loop

```python
def start_and_ending_position_brute(nums, target):
    start = -1
    end = -1

    for i in range(len(nums)):
        if nums[i] == target:
            if start == - 1:
                start = i
            end = i

    return (start, end)
```

Alright, we were successfully able to implement the brute force approach using linear search and we have a time complexity of O(N) and space complexity of O(1). We can optimize this further by implementing the binary search approach

Here, the steps involved:
1. Split the list `nums` in half and get the `mid` position and `middle element`
2. Check if the `middle element` is equal to the `target`, if yes,
3. Check if  `middle element` is equal to the number before it
4. If yes, move `left` and repeat till it doesn't match again else, return that `position`
5. Again, check if the `middle element` is equal to the number after it
6. If yes, move `right` and repeat till it doesn't match again
7. Return the `position` of the last occurrence

Here, let's use two functions instead of one: `starting_position` and `ending_position`

```python
def binary_search(lo, hi, condition):
    mid = (lo + hi) // 2
    result = condition(mid)
    while lo <= hi:
        if result == 'found':
            return mid
        elif result == 'left':
            hi = mid - 1
        elif result == 'right':
            lo = mid + 1
    return -1    
```

```python
def starting_position(nums, target):

    def condition(mid):
        if nums[mid] == target:
            if mid - 1 >= 0 and nums[mid] == nums[mid - 1]:
                return 'left'
            return 'found'
        
    return binary_search(-1, -1, condition)
```
