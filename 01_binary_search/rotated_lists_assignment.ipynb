## Count Rotations in a Rotated Sorted Array
Given an integer array nums of length $n$ sorted in ascending order with distinct values, suppose it has been rotated between $0$ and $n-1$ times.

A single rotation shift moves the last element of the array to the front (right rotation). For example, rotating [1, 2, 3, 4, 5] 3 times results in [3, 4, 5, 1, 2].

Return the total number of times the original array was rotated.

You must write an algorithm that runs in $O(\log n)$ time complexity.

## Problem Statement

Write a function `count_rotations` to determine how many times an ascending sorted list of numbers was rotated.

## Input
1. `nums`: A list of numbers that was originally sorted in ascending order and then rotated (e.g., `[5, 6, 7, 1, 2, 3, 4]`).

## Output

1. `rotations`: An integer representing the number of times the list was rotated (e.g., `3`).

## Test Cases

1. Array rotated somewhere in the middle. 
2. Array not rotated.
3. Array with just one number.
4. Smallest element of array at the very end.
5. Array is empty.
