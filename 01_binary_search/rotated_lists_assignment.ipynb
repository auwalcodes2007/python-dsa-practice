## QUESTION

Search in Rotated Sorted Array (LeetCode #33)

There is an integer array nums sorted in ascending order with distinct values.

Prior to being passed to your function, nums is possibly rotated at an unknown pivot index $k$ ($1 \le k < \text{nums.length}$) such that the resulting array is $[nums[k], nums[k+1], \dots, nums[n-1], nums[0], nums[1], \dots, nums[k-1]]$ (0-indexed).

Example of Rotation:The array [0, 1, 2, 4, 5, 6, 7] rotated at pivot index 3 becomes [4, 5, 6, 7, 0, 1, 2].

Given the array nums after the possible rotation and an integer target, return the index of target if it is in nums, or -1 if it is not in nums.

You must write an algorithm with $O(\log n)$ runtime complexity.
