## Let's come up with a brute solution
1. Set an empty variable `rotations` to `0` 
2. Loop through the list `nums`
3. Find the smallest term
4. Return the index of the smallest term
5. If no terms, return `0`

```python
def count_rotations_brute(nums):
    rotations = 0

    for i in range(0, len(nums) - 1):
        if nums[i] > nums[i + 1]:
            rotations = i + 1            

    return rotations
```

## Optimized Solution
Alright, that was the brute force solution with a complexity of $O(N)$, let's optmize it further using binary search

Steps for implementation
1. Split list in half, at position `mid`
2. Check if the number at the `mid` position is greater than number at the next position
3. If yes, return the next position as the number of `rotations`
4. Else, check if the number at the `mid` position is greater than the number at position `hi`
5. If it is greater than it, then return `right`, else, return `left`



```python
def binary_search(lo, hi, condition):
    while lo <= hi:
        mid = (lo + hi) // 2
        result = condition(mid)
        if result == "found":
            return mid
        elif result == "left":
            hi = mid - 1
        else: 
            lo = mid + 1
            
    return 0
```

```python
def count_rotations_binary(nums):
    lo = 0
    hi = len(nums) - 1

    def condition(mid):
        if mid - 1 >= 0 and nums[mid] < nums[mid - 1]:
            return "found"
        elif nums[mid] > nums[hi]:
            return "right"
        else: 
            return "left"

    return binary_search(lo, hi, condition)
```
