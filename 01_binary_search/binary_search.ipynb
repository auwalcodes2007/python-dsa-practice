## General Binary Search

1. Come up with a condition to determine where the answer lies, whether before, after or at the given position
2. Retrieve the midpoint and middle element of the list.
3. If it is the answer, return the middle position as the answer.
4. If the answer lies before it, repeat the search for the first half of the list.
5. If the answer lies after it, repeat the search for the second half of the list.

Here's the implementation in python:

```python
def binary_search(lo, hi, condition):
    """"TODO - add docs"""
    while lo <= hi:
        mid = (lo + hi) // 2
        result = condition(mid)
        if result == 'found':
            return mid
        elif result == 'left':
            hi = mid - 1 
        elif result == 'right':
            lo = mid + 1
    return -1
```

The worst-case or running time of binary search is O(log N) provided the complexity of the condition used to determine whether the answer lies before, after or at a given position is O(1).
