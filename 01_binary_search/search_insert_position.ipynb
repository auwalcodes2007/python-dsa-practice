## Pseudocode

1. Initialize `lo = 0` and `hi = len(nums) - 1`
2. Compute `mid = (lo + hi) // 2`.
2. If `nums[mid] == target`, return `mid`
3. if `nums[mid]` < target, move right `(lo = mid + 1)`
4. If `nums[mid]` > target, move left `(hi = mid - 1)`.

```python
def search_insert_position(nums, target):
    lo, hi, mid = 0, len(nums) - 1, 0

    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            lo = mid + 1
        else: 
            hi = mid - 1

    return lo
```
