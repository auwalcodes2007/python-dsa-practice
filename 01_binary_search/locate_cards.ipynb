```python
# query occurs in the middle
tests.append(
    {
        "input": {
            "cards": [13, 11, 10, 7, 4, 3, 1, 0],
            "query": 1
        },
        "output": 6
    }
)
```

```python
# query is first element
tests.append(
    {
        "input": {
            "cards": [4, 2, 1, -1],
            "query": 4
        },
        "output": 0
    }
)
```

```python
# query is last element
tests.append(
    {
        "input": {
            "cards": [3, -1, 9, -127],
            "query": -127
        },
        "output": 3
    }
)
```

```python
# cards contains just one element, query
tests.append(
    {
        "input": {
            "cards": [6],
            "query": 6
        },
        "output": 0
    }
)
```

```python
# card doesn't contain query
tests.append(
    {
        "input": {
            "cards": [13, 11, 10, 7, 4, 3, 1, 0],
            "query": 9
        },
        "output": -1
    }
)
```

```python
# cards is empty
tests.append(
    {
        "input": {
            "cards": [],
            "query": 1
        },
        "output": -1
    }
)
```

```python
# numbers can repeat in cards
tests.append(
    {
        "input": {
            "cards": [8, 8, 8, 6, 6, 3, 2, 2, 2, 2, 0, 0, 0],
            "query": 3
        },
        "output": 5
    }
)
```

```python
# query occurs multiple times 
tests.append(
    {
        "input": {
            "cards": [8, 8, 8, 6, 6, 3, 2, 2, 2, 2, 0, 0, 0],
            "query": 6
        },
        "output": 3
    }
)
```

```python
tests
```

```python
def locate_card(cards, query):
    position = 0

    while True:
        if cards[position] == query: 
            return position

        position += 1
        
        if position == len(cards):
            return -1
```
