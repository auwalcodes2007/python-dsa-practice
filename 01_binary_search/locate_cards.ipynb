## 5. Analyze algorithm's complexity and identify inefficiencies
As seen above, we have successfully implemented the linear search algorithm to solve this problem. But, we are having a `O(N)` time complexity since the time is proportional to the size of the array i.e time grows with input size AND a `O(1)` space complexity. Although, we can solve this problem using binary search which would reduce our runtime. Here, the search is started from the middle of the list and checks if the `query` is greater than or less than the `middle` value, it then discards the other half of the list which isn't required and repeats the process till we arrive at the desired output.

Here are the steps to implementing this solution
1. Set initial values for `lo` and `hi`. Where `lo` is the first index in the list and `hi` and is the last index in the list
2. Calculate the mid position of the list using floor/integer division (since we are dealing with whole number list indexes) and determine the `middle` number.
3. If the `middle` number is equal to the `query`, then return the `mid` position
4. Else if the `middle` number is less than the `query`, then set the `lo` to one index higher than the `mid` position.
5. Else if the `middle` number is greater than the `query`, then set the `hi` to one index lower than the `mid` position.
6. If the number isn't found, return `-1`

```python
def locate_card(cards, query):
    lo, hi, mid = 0, len(cards) - 1, 0

    while lo <= hi:
        mid = (lo + hi) // 2
        middle_number = cards[mid]
        if query == middle_number:
            return mid
        elif query < middle_number:
            lo = mid + 1
        elif query > middle_number:
            hi = mid - 1
        
    return -1
```

As we can see, all the tests pass except the one where we have duplicates of the `query` in the list `cards`. This is because if we select a value "in random", we may get the `query` but not the first appearance and our test cases expect exactly the first occurrence of that `query`


Thus we need to write a helper function to help determine whether to go leftwards even after finding the `query` in order to ensure there are no earlier duplicates.

```python
def test_location(cards, query, mid):
    mid_number = cards[mid]
    if mid_number == query:
        if mid - 1 >= 0 and query == cards[mid - 1]:
            return "left"
        else:
            return "found"        
    elif mid_number < query:
        return "left"
    elif mid_number > query: 
        return "right"
```

Now let's rewrite the locate_card function with this helper function we just came up with

```python
def locate_card(cards, query):
    lo, hi, mid = 0, len(cards) - 1, 0

    while lo <= hi:
        mid = (lo + hi) // 2
        search_direction = test_location(cards, query, mid)
        if search_direction == "found":
            return mid
        elif search_direction == "right":
            lo = mid + 1
        elif search_direction == "left":
            hi = mid - 1
        
    return -1
```
