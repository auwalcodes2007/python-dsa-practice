Here is a quick refresher of classes and objects in python. We want to represent a user's profile containing username, name and email using python classes

```python
class User:
    pass
```

```python
user1 = User()
```

Let's define our User profile

```python
class User:
    def __init__(self, username, name, email):
        self.username = username
        self.name = name
        self.email = email

    def introduce_yourself(self, guest_name="Joe"):
        print("Hi {}, I am {}! Contact me at {}".format(guest_name, self.name, self.email))
```

```python
user2 = User("auwal", "Auwal Hassan", "auwalhassan@email.com")
user2.introduce_yourself()
```

The above is the same as:

```python
User.introduce_yourself(user2)
```
