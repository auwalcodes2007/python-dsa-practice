Let's define a couple of helper functions in our class for better representation of it

```python
class User:
    def __init__(self, username, name, email):
        self.username = username
        self.name = name
        self.email = email

    def introduce_yourself(self, guest_name="Joe"):
        print("Hi {}, I am {}! Contact me at {}".format(guest_name, self.name, self.email))

    def __repr__(self):
        return f"User(username='{self.username}', name='{self.name}', email='{self.email}')"

    def __str__(self):
        return f"User {self.username}"
```

## Exercise 1: 
What is the purpose of defining the functions __str__ and __repr__ within a class? How are the two functions different? Illustrate with some examples below.

## Solution:

We define __str__ and __repr__ in order to represent our objects in an easy to understand format so we can easily differentiate between objects of classes.

__repr__ is the what I would call 'developer/debugging' lower level representation of the object. Why do I say so? Well, when an object is called, the python interpreter represents that object with the value from __repr__, whereas where it is not provided, the memory location is provided instead. 
Whereas, __str__ is the 'client/end user' higher level representation of the object. Here, the object can be represented in a closer to plain English format for easier understanding. It is normally called when the object is passed in the `print()` function or `str()` function. If not provided, it defaults to the __repr__

Below is an illustration using a book analogy:

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"

    def __str__(self):
        return f"Book {self.title} by {self.author}"
```

```python
book = Book('Harry Potter', 'Auwal')
```

For the __repr__, you can see it in action here:

```python
book 
```

For the __str__, here it is in action:

```python
print(book)
str(book)
```
