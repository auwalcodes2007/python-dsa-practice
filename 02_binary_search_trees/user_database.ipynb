## Solution:

We define __str__ and __repr__ in order to represent our objects in an easy to understand format so we can easily differentiate between objects of classes.

__repr__ is the what I would call 'developer/debugging' lower level representation of the object. Why do I say so? Well, when an object is called, the python interpreter represents that object with the value from __repr__, whereas where it is not provided, the memory location is provided instead. 
Whereas, __str__ is the 'client/end user' higher level representation of the object. Here, the object can be represented in a closer to plain English format for easier understanding. It is normally called when the object is passed in the `print()` function or `str()` function. If not provided, it defaults to the __repr__

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"

    def __str__(self):
        return f"Book {self.title} by {self.author}"
```

## Come up with a correct solution. State it in plain English.

The various functions can be implemented as follows:
1. Insert: Loop through the list and add the new user at a position that keeps the list sorted.
2. Find: Loop through the list and find the user object with the username matching the query.
3. Update: Loop through the list and find the user object matching the query and update the details.
4. List: Return the list of user objects.

## Implement the solution and test with example inputs

```python
class UserDatabase:
    def __init__(self):
        self.users = []

    def insert(self, user):
        i = 0
        while i < len(self.users):
            if self.users[i].username > user.username:
                break
            i += 1
        self.users.insert(i, user)

    def find(self, username):
        for user in self.users:
            if user.username == username:
                return user
        return None

    def update(self, user):
        target = self.find(user.username)
        target.name, target.email = user.name, user.email

    def list_all(self):
        return self.users
```

## Analyze the algorithm's complexity and identify inefficiencies

The operations `insert`, `find` and `update` involve iterating over a list of users, and in the worst case, they may take up `N iterations` to return a result, where `N` is the total number of users. `list_all` however, simply returns the existing internal list of users.

Thus, the time complexity of the various operations are: 
* Insert: $O(N)$
* Find: $O(N)$
* Update: $O(N)$
* List: $O(1)$

**Exercise**: Verify that the space complexity of each operation is $O(1)$

**Attempt**: The space complexity of each operation is $O(1)$ because the extra memory allocated by the operation stays consistent, regardless of how many users $(N)$ are in the database.

But this is not optimal. To get a sense of how long each function may take if there were 100 million users on the platform, we can simply run a for or while loop on 10 million numbers

```python
%%time
for i in range(100000000):
    j = i * i
```

As seen above, it takes almost 51 seconds to execute all iterations in that cell
* A 51-second delay for fetching user profiles will lead to a suboptimal users experience and may cause many users to stop using the platform altogether
* A 51-second processing time for each profile request would also significantly reduce the number of users that can access the platform at a time or increase cloud infrastructure costs for the company by millions of dollars. Simply, this is because, imagine you have 8 CPU cores on a backend computer server and a core attends to one user at a time and 8 users try to access their profiles(`find`) then the cores can only attend to 8 users at a time for 51 seconds which is not good at all.

Technically: An 8-core server handling 8 blocking linear scan requests will pin all 8 CPU cores at 100% capacity for the full duration of those requests. Any 9th user trying to connect gets queued or dropped because no execution threads remain open to process their HTTP connection

**Note**: `user` represents time taken by CPU whether single or multi thread. Whereas, `wall time` represents time elapsed in the real world

Thus, we need to come up with a more optimal solution.

We can solve this problem by organising our data in a structure called **Binary Trees**

### **Binary Tree**
A **binary tree** is a hierarchical, non-linear data structure where **each node can have a maximum of two child nodes**(0, 1, 2 i.e. why it is called binary), traditionally referred to as the **left child** and the **right child**. 

### Core Terminology
- **Root**: The topmost node of the tree, which has no parent. This is where operations like search, insertion, etc. begin
- **Parent / Child:** A node that points to another node is its parent, and the connected node below it is the child.
- **Leaf Node:** A node at the very bottom that has no children.

For our use case, we require the binary tree to have additional properties:
1. **Keys and Values:** Each node would represent a key(a username) and a value(User object). A binary tree where nodes contain both a key and a value is referred to as a **map** or **treemap**(because it maps keys to values)

2. **Binary Search Tree:** The *left subtree* of any node contains nodes with keys that are lexicographically smaller than the node's key, and the *right subtree* of any node contains nodes with keys that are lexicographically larger than the node's key. This makes it easier to locate a specific key by traversing a single path down from the root node.

3. **Balanced Tree:** The tree does not skew too heavily to one side or the other. The left and right subtrees of any node should not differ in height/depth by more than 1 level.
