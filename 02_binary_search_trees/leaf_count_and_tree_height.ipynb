```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
```

```python
tree_tuple = ((1, 3, None), 2, (((None, 3, 4), 5, (6, 7, 8))))
```

```python
def tuple_to_tree(data):
    if data is None:
        return None

    if not isinstance(data, tuple):
        return TreeNode(data)

    if len(data) != 3:
        raise ValueError("Tuple must have exactly 3 elements: (left, key, right)")

    left, key, right = data
    node = TreeNode(key)
    node.left = tuple_to_tree(left)
    node.right = tuple_to_tree(right)

    return node
```

```python
tree = tuple_to_tree(tree_tuple)
```

```python
def count_leaves(node):
    
    if node is None:
        return 0
    
    if node.left is None and node.right is None:
        return 1

    return count_leaves(node.left) + count_leaves(node.right)
```

```python
count_leaves(tree)
```

```python
def tree_height(node):
    if node is None:
        return 0

    return 1 + max(tree_height(node.left), tree_height(node.right))
```

### Note: 
We used an node based count for the tree height, if we wanted an edge based, we would return -1 where `node is None`

```python
tree_height(tree)
```
