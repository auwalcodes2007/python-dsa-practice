```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
```

```python
tree_tuple = ((1, 3, None), 2, (((None, 3, 4), 5, (6, 7, 8))))
```

```python
def tuple_to_tree(data):
    if data is None:
        return None

    if not isinstance(data, tuple):
        return TreeNode(data)

    if len(data) != 3:
        raise ValueError("Tuple must have exactly 3 elements: (left, key, right)")

    left, key, right = data
    node = TreeNode(key)
    node.left = tuple_to_tree(left)
    node.right = tuple_to_tree(right)

    return node
```
