### **Binary Tree**
A **binary tree** is a hierarchical, non-linear data structure where **each node can have a maximum of two child nodes**(0, 1, 2 i.e. why it is called binary), traditionally referred to as the **left child** and the **right child**. 

### Core Terminology
- **Root**: The topmost node of the tree, which has no parent. This is where operations like search, insertion, etc. begin
- **Parent / Child:** A node that points to another node is its parent, and the connected node below it is the child.
- **Leaf Node:** A node at the very bottom that has no children.

For our use case, we require the binary tree to have additional properties:
1. **Keys and Values:** Each node would represent a key(a username) and a value(User object). A binary tree where nodes contain both a key and a value is referred to as a **map** or **treemap**(because it maps keys to values)

2. **Binary Search Tree:** The *left subtree* of any node contains nodes with keys that are lexicographically smaller than the node's key, and the *right subtree* of any node contains nodes with keys that are lexicographically larger than the node's key. This makes it easier to locate a specific key by traversing a single path down from the root node.

3. **Balanced Tree:** The tree does not skew too heavily to one side or the other. The left and right subtrees of any node should not differ in height/depth by more than 1 level.

### Height of a tree
The number of levels in a tree is called its height. Each level of a tree contains twice as many nodes as the previous level.

For a tree of height `k`, here is a list of the number of nodes at each level:

Level 0: `1`

Level 1: `2`

Level 2: `4` i.e $`2^2`$

$\dots$ 

Level k-1: $`2^{k-1}`$

If the total number of nodes in the tree is `N`, it follows that:

`N` = 1 + $`2^1`$ + $`2^2`$ + $\dots$ + $`2^{k-1}`$

We can simplify by adding `1`

`N` + 1 = 1 + 1 + $`2^1`$ + $`2^2`$ + $\dots$ + $`2^{k-1}`$

`N` + 1 = $`2^1`$ + $`2^1`$ + $`2^2`$ + $\dots$ + $`2^{k-1}`$

`N` + 1 = $`2^2`$ + $`2^2`$ + $\dots$ + $`2^{k-1}`$

$\dots$

`N` + 1 = $`2^{k-1}`$ + $`2^{k-1}`$

`N` + 1 = $`2^k`$

k = log(N + 1) <= log(N) + 1

**Quick Note:** This can also be gotten using sum of series in a G.P. 

$$N = 2^0 + 2^1 + 2^2 + \dots + 2^{k-1}$$

Using the geometric series summation formula $S_n = \frac{a(r^k - 1)}{r - 1}$ (where first term $a = 1$ and common ratio $r = 2$):

$$N = \frac{1(2^k - 1)}{2 - 1}$$

$$N = 2^k - 1$$

Implement a binary tree with Python, and show its usage with some examples.

### Solution
Here's a simple class representing a node in a Binary Tree

```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
```

As seen above, by design, our node is independent of other nodes upon instantiation.

```python
node0 = TreeNode(3)
node1 = TreeNode(4)
node2 = TreeNode(5)
```

We can connect the nodes by setting the `.left` and `.right` properties of the root node

```python
node0.left = node1
node0.right = node2
```

```python
node0.left, node0.right
```

```python
node1, node2
```

Now we may want to track which node is the `root` node and use it to access all nodes within the tree. Here, `tree` points to the `root` node

```python
tree = node0
```

```python
tree.key, tree.left.key
```

That's the solution to the problem. Going forward, the term `tree` would be used to refer to the `root` node. While the term `node` would be used to refer to any node in the tree, not necessarily the `root` node.

### Exercise(Check youtube video at 3:16:47 for diagram): 
Create the following binary tree using the TreeNode class just defined 

```python
tree_tuple = ((1, 3, None), 2, (((None, 3, 4), 5, (6, 7, 8))))
```

Let's create a function to take values. Where `tree_tuple = (left_subtree, key, right_subtree)`

```python
def parse_tuple(data):
    if data is None:
        return None

    if not isinstance(data, tuple):
        return TreeNode(data)

    left_leaf, key, right_leaf = data

    node = TreeNode(key)
    node.left = parse_tuple(left_leaf)
    node.right = parse_tuple(right_leaf)

    return node
    
```

```python
tree2= parse_tuple(tree_tuple)
```

```python
tree2.key, tree2.left.right
```

**Exercise**: Define a function `tree_to_tuple` that converts a binary tree into a tuple representing the same tree. E.g. `tree_to_tuple` converts the tree created above to the tuple `((1, 3, None), 2, (((None, 3, 4), 5, (6, 7, 8))))`. *Hint:* Use recursion.

```python
def tree_to_tuple(tree):
    if tree is None:
        return None

    if tree.left is None and tree.right is None:
        return tree.key

    return (tree_to_tuple(tree.left), tree.key, tree_to_tuple(tree.right))
```

```python
tree_to_tuple(tree2)
```

We can create a funtion to display the keys

```python
def display_keys(node, space='  ', level=0):
    # Base Case: Empty node
    if node is None:
        print(space*level + '∅')
        return

    # Base Case: Leaf node
    if node.left is None and node.right is None:
        print(space*level + str(node.key))
        return

    # Recursive Step: Right -> Root -> Left (Reversed In-Order)
    display_keys(node.right, level+1)
    print(space*level + str(node.key))
    display_keys(node.left, level+1)
    
```

If you flip the result  90 degrees when the function is called in a clockwise direction, you would see the tree shape better.

```python
display_keys(tree2)
```

### Traversing a Binary Tree

A *traversal* refers to the process of visiting each node of a binary tree exactly once. *Visiting a node* generally refers to adding the node's key to a list. There are three ways to traverse a binary tree and return a list of visited keys:

#### Inorder traversal
1. Traverse the left subtree recursively inorder
2. Traverse the current node
3. Traverse the right subtree recursively inorder


#### Preorder traversal
1. Traverse the current node
2. Traverse the left subtree recursively preorder
3. Traverse the right  subtree recursively preorder


#### Postorder traversal
1. Traverse the left subtree recursively postorder
2. Traverse the right subtree recursively postorder
3. Traverse the current node

Alright let's write a function to traverse the tree inorder.

```python
def traverse_in_order(node):
    if node is None:
        return []

    print(traverse_in_order(node.left) + [node.key] + [node.right])
```

```python
traverse_in_order(tree2)
```
